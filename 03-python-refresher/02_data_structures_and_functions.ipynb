## 📌 Topic 3.3: Lists & List Comprehensions

📌 **Definition:** A List is an ordered, mutable collection of items enclosed within square brackets `[]`.

```python
# Prices in USD
prices_usd = [12.5, 45.0, 8.99, 120.0, 25.5]
usd_to_inr = 83.0

# 1-Line List Comprehension: Convert items > $10 to INR
prices_inr = [round(p * usd_to_inr, 2) for p in prices_usd if p > 10.0]
print("Converted INR Prices:", prices_inr)
```

## 📌 Topic 3.4: Reusable Functions (`def`)

📌 **Definition:** Functions are named, reusable blocks of code that perform specific calculations.

```python
def calculate_invoice(base_price, gst_pct=18, discount=0):
    discounted = max(0, base_price - discount)
    tax = discounted * (gst_pct / 100)
    return discounted + tax

print("Laptop Invoice:", calculate_invoice(50000, discount=2000))
print("Book Invoice:", calculate_invoice(500, gst_pct=5))
```
