## 📌 Topic 3.2: Control Flow (`if`, `elif`, `else`)

📌 **Definition:** Control flow evaluates boolean conditions to direct execution paths.

```python
# Customer Loyalty Tier Classification
annual_spend = 75000

if annual_spend >= 100000:
    tier = "Platinum"
elif annual_spend >= 50000:
    tier = "Gold"
else:
    tier = "Standard"

print(f"Customer Loyalty Tier: {tier}")
```
