## 📌 1. Theory & Intuition
- **5-Step EDA Framework:** 1. Head & Info ➔ 2. Missing/Duplicates ➔ 3. Summary Stats ➔ 4. Feature Distributions ➔ 5. Correlation Heatmap.
- **Multicollinearity:** Identifying highly correlated features ($r > 0.8$) before training linear models.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# 💻 Live Demo: Loading Dataset & Plotting Correlation Heatmap
url = "https://raw.githubusercontent.com/SahilShaikh-py/Advanced-Python-Data-Science-/main/datasets/housing_prices.csv"
df = pd.read_csv(url)

print("--- Step 1: Dataset Summary Statistics ---")
print(df.describe())

# Compute Pearson Correlation Matrix
corr_matrix = df.drop(columns=["House_ID"]).corr()

# Plot Seaborn Heatmap
plt.figure(figsize=(8, 6))
sns.heatmap(corr_matrix, annot=True, cmap="coolwarm", fmt=".2f", linewidths=0.5)
plt.title("Housing Features Correlation Matrix Heatmap")
plt.show()
```
