## 1. Definition
- Exploratory Data Analysis (EDA) systematically inspects dataset distributions, anomalies, missing data, and feature relationships.
- Visual Storytelling translates numerical correlation matrices into intuitive heatmaps and scatterplots.

## 2. Key Features
- 5-Step EDA framework covers data inspection, cleaning, summary stats, distributions, and correlations.
- Seaborn provides Statistical Data Visualization themes and automated distribution plots.
- Pearson Correlation Coefficient r measures linear association between numerical features.
- Multicollinearity detection identifies collinear features prior to machine learning model fitting.

## 3. Real-World Uses
- Auditing raw enterprise database tables before machine learning training.
- Communicating data insights and feature correlations to business stakeholders.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Live Instructor Code-Along: 5-Step EDA Pipeline
url = "https://raw.githubusercontent.com/SahilShaikh-py/Advanced-Python-Data-Science-/main/datasets/housing_prices.csv"
df = pd.read_csv(url)

print("=== Step 1: Data Types and Null Count ===")
print(df.info())

print("
=== Step 2: Summary Statistics ===")
print(df.describe())

# Compute Correlation Matrix
corr_matrix = df.drop(columns=["House_ID"]).corr()

# Plot Seaborn Heatmap
plt.figure(figsize=(8, 6))
sns.heatmap(corr_matrix, annot=True, cmap="coolwarm", fmt=".2f", linewidths=0.5)
plt.title("Housing Features Correlation Matrix Heatmap")
plt.show()
```
