## 1. Definition
- Hypothesis Testing evaluates whether observed sample differences reflect true population metrics or random noise.
- A/B Testing compares a Control group (A) against a Treatment group (B) to prove statistical significance.

## 2. Key Features
- Null Hypothesis (H0) assumes no difference between sample groups.
- Alternative Hypothesis (H1) posits a statistically significant difference.
- p-value quantifies probability under H0; if p-value <= 0.05, H0 is rejected.
- Central Limit Theorem states sample means approach a Gaussian normal distribution.

## 3. Real-World Uses
- Testing digital checkout UI feature variations.
- Measuring marketing campaign click-through rate improvements.

```python
import numpy as np
from scipy import stats

# Live Instructor Code-Along: Two-Sample Independent t-Test
np.random.seed(42)

# Group A (Control Original Checkout Page): Mean spend $48.50
group_a = np.random.normal(loc=48.50, scale=8.0, size=300)

# Group B (Variant 1-Click Checkout Page): Mean spend $52.10
group_b = np.random.normal(loc=52.10, scale=8.0, size=300)

t_stat, p_val = stats.ttest_ind(group_a, group_b)

print(f"Group A Mean Spend: ${np.mean(group_a):.2f}")
print(f"Group B Mean Spend: ${np.mean(group_b):.2f}")
print(f"t-Statistic: {t_stat:.4f} | p-value: {p_val:.6f}")

if p_val < 0.05:
    print("Status: REJECT Null Hypothesis. Variant B increased spend significantly.")
else:
    print("Status: FAIL to Reject Null Hypothesis. No statistical difference.")
```
