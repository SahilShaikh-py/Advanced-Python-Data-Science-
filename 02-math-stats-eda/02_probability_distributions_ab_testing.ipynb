## 📌 1. Theory & Intuition
- **Central Limit Theorem (CLT):** The sample distribution of means approaches a Normal Gaussian distribution as sample size $N \ge 30$.
- **A/B Testing Framework:** Compare Control A vs Variant B. If $p\text{-value} \le 0.05$, reject $H_0$ and confirm statistically significant uplift.

```python
import numpy as np
from scipy import stats

# 💻 Live Demo: Conducting a 2-Sample Independent t-Test for A/B Testing
np.random.seed(42)
# Control A (Original Checkout Page): Mean spend $50
group_a = np.random.normal(loc=50.0, scale=10.0, size=250)

# Variant B (1-Click Buy Checkout Page): Mean spend $53
group_b = np.random.normal(loc=53.0, scale=10.0, size=250)

# Run Two-Sample Independent t-Test
t_stat, p_val = stats.ttest_ind(group_a, group_b)

print(f"Group A Mean Spend: ${np.mean(group_a):.2f}")
print(f"Group B Mean Spend: ${np.mean(group_b):.2f}")
print(f"t-Statistic: {t_stat:.4f} | p-value: {p_val:.6f}")

if p_val < 0.05:
    print("✅ Statistical Significance Reached! Variant B increased spend significantly.")
else:
    print("❌ Difference is not statistically significant.")
```
