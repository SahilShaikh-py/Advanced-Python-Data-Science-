## 📌 1. Theory & Intuition
- **Vector Space & Cosine Similarity:** Cosine of the angle between two vectors measures direction similarity ($1.0 = \text{Identical}, 0.0 = \text{Orthogonal}$).  
- **Gradient Descent:** Iteratively updates parameters along the negative gradient of the loss function: $w = w - \alpha \nabla L$.

```python
import numpy as np

# 💻 Live Demo: Writing Gradient Descent from Scratch in Python
# Objective: Find weight 'w' that minimizes Mean Squared Error Loss L(w) = (w - 7)^2
# Derivative dL/dw = 2 * (w - 7)

w = 0.0         # Initial weight
learning_rate = 0.1
epochs = 15

print("Starting Gradient Descent Optimization:")
for epoch in range(epochs):
    gradient = 2 * (w - 7)                   # Calculate gradient
    w = w - (learning_rate * gradient)       # Apply update rule
    loss = (w - 7) ** 2                      # Compute loss
    print(f"Epoch {epoch+1:02d}: w = {w:.4f} | Loss = {loss:.4f}")

print(f"\nOptimal Weight Found: {w:.2f} (Target: 7.00)")
```
