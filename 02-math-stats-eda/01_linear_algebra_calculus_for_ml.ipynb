## 1. Definition
- Linear Algebra provides the mathematical framework for vector spaces, projections, and matrix transformations.
- Differential Calculus calculates partial derivatives to steer gradient descent algorithms downhill.

## 2. Key Features
- Cosine Similarity measures the angular alignment between multi-dimensional vectors.
- Partial derivatives determine the rate of loss change with respect to individual weights.
- Learning rate alpha scales the parameter step size per optimization step.
- Vector norms (L1, L2) quantify vector lengths for model regularization.

## 3. Real-World Uses
- Training parameters in Linear Regression, Logistic Regression, and Deep Neural Networks.
- Dimensionality reduction using Eigenvalue decomposition in PCA.
- Cosine similarity search in recommendation engines.

```python
import numpy as np

# Live Instructor Code-Along: Gradient Descent Algorithm from Scratch

# Goal: Minimize Loss Function L(w) = (w - 7.5)^2
# Partial derivative dL/dw = 2 * (w - 7.5)

w = 0.0          # Initial weight
learning_rate = 0.15
epochs = 20

print("Optimization Progress:")
for epoch in range(epochs):
    gradient = 2 * (w - 7.5)                   # Derivative computation
    w = w - (learning_rate * gradient)         # Gradient Descent update rule
    loss = (w - 7.5) ** 2                      # Loss evaluation
    print(f"Epoch {epoch+1:02d}: Weight = {w:.4f} | Loss = {loss:.4f}")

print(f"Final Converged Weight: {w:.2f} (Target: 7.50)")
```
