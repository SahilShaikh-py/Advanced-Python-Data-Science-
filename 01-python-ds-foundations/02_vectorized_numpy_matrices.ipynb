## 📌 1. Theory & Intuition
- **Contiguous Memory:** NumPy stores array elements in adjacent C memory buffers, enabling CPU SIMD hardware acceleration.
- **Broadcasting:** Automatic alignment of smaller vectors across 2D/3D matrix dimensions.
- **Matrix Multiplication (`@` or `np.dot`):** Fundamental operation for neural network layers.

```python
import numpy as np

# 💻 Live Demo: 2D Matrix Multiplication & Feature Scaling
# 4 samples, 3 features (Height, Weight, Age)
features = np.array([
    [170, 65, 25],
    [180, 80, 30],
    [160, 55, 22],
    [175, 70, 28]
], dtype=np.float64)

# 1. Compute Feature Means along columns (axis=0)
means = np.mean(features, axis=0)
stds = np.std(features, axis=0)

# 2. Vectorized Mean Centering via Broadcasting
scaled_features = (features - means) / stds

# 3. Matrix Multiplication: Linear combination using Weight vector [0.5, 0.3, 0.2]
weights = np.array([0.5, 0.3, 0.2])
predictions = np.dot(scaled_features, weights)

print("Feature Means:", means)
print("Scaled Features Matrix:
", np.round(scaled_features, 2))
print("Calculated Predictions:", np.round(predictions, 2))
```
