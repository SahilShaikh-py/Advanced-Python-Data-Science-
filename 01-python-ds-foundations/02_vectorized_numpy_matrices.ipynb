## 1. Definition
- Vectorization executes operations on whole multi-dimensional arrays simultaneously using underlying compiled C routines.
- NumPy arrays allocate homogeneous data elements in contiguous memory buffers.

## 2. Key Features
- Contiguous memory layout maximizes CPU cache hit rates and SIMD instructions.
- Broadcasting rules align lower-dimensional arrays across higher dimensions during operations.
- Reshaping operations change array dimensions without copying underlying memory.
- Linear algebra routines (`np.dot`, `np.linalg.inv`) compute fast matrix transformations.

## 3. Real-World Uses
- Neural network layer weight transformations and activation computations.
- Image pixel grid transformations and filter convolutions.
- Portfolio returns and covariance matrix computations.

```python
import numpy as np

# Live Instructor Code-Along: 2D Matrix Manipulation, Scaling and Multiplication

# 1. Create a 5x3 Feature Matrix (5 customers, 3 features: Spend, Age, Support_Calls)
features = np.array([
    [1200.0, 34.0, 2.0],
    [450.0,  22.0, 5.0],
    [3100.0, 45.0, 0.0],
    [890.0,  29.0, 1.0],
    [2100.0, 38.0, 3.0]
], dtype=np.float64)

# 2. Vectorized Feature Standardization (Mean = 0, Std = 1)
means = np.mean(features, axis=0)
stds = np.std(features, axis=0)
scaled_features = (features - means) / stds

# 3. Linear Model Weights vector [0.4, 0.3, -0.5]
weights = np.array([0.4, 0.3, -0.5])

# 4. Matrix-Vector Multiplication (Dot Product)
churn_scores = np.dot(scaled_features, weights)

print("Column Means:", np.round(means, 2))
print("Scaled Matrix Shape:", scaled_features.shape)
print("Calculated Churn Scores:", np.round(churn_scores, 3))
```
