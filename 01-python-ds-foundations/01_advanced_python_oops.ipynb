## 1. Definition
- Object-Oriented Programming (OOP) structures code into reusable Class blueprints and instance Objects.
- Generators use the `yield` statement to return a lazy iterator stream without loading entire data structures into RAM.

## 2. Key Features
- Encapsulation protects internal state variables while providing clean methods like `fit()` and `transform()`.
- Inheritance enables child data transformers to inherit common methods from abstract base classes.
- Dunder methods like `__len__`, `__getitem__`, and `__repr__` integrate custom classes with native Python functions.
- Lazy evaluation in generators conserves system memory during big data processing.

## 3. Real-World Uses
- Building custom data preprocessing pipelines compatible with Scikit-Learn.
- Streaming large multi-gigabyte CSV or JSON logs line-by-line.
- Packaging production machine learning models as OOP Python modules.

```python
# Live Instructor Code-Along: Advanced Preprocessing Class with Stream Generator

class BaseTransformer:
    '''Base abstract transformer class for data science pipelines'''
    def __init__(self, feature_name):
        # Explanation: Store feature identifier
        self.feature_name = feature_name
        self.is_fitted = False

class StandardScaler(BaseTransformer):
    '''Custom Z-Score Standard Scaler implementing fit and transform'''
    def __init__(self, feature_name):
        super().__init__(feature_name)
        self.mean = 0.0
        self.std = 1.0

    def fit(self, data_stream):
        # Explanation: Compute mean and standard deviation from data list
        data = list(data_stream)
        self.mean = sum(data) / len(data)
        variance = sum((x - self.mean) ** 2 for x in data) / len(data)
        self.std = variance ** 0.5 if variance > 0 else 1.0
        self.is_fitted = True
        print(f"[FIT COMPLETE] Feature: {self.feature_name} | Mean: {self.mean:.2f} | Std: {self.std:.2f}")
        return self

    def transform(self, data_stream):
        # Explanation: Apply Z-Score formula (x - mean) / std
        if not self.is_fitted:
            raise RuntimeError("Call fit before transform!")
        for x in data_stream:
            yield (x - self.mean) / self.std

# Generator Function for Memory-Efficient Data Generation
def generate_salary_data(n=1000):
    # Explanation: Yield synthetic salary numbers lazily
    import random
    for _ in range(n):
        yield random.gauss(60000, 15000)

# Test Pipeline
scaler = StandardScaler(feature_name="Annual_Salary")
raw_stream = list(generate_salary_data(500))
scaler.fit(raw_stream)

# Transform first 5 items
transformed_stream = scaler.transform(raw_stream[:5])
print("Transformed Z-Scores:", [round(z, 3) for z in transformed_stream])
```
