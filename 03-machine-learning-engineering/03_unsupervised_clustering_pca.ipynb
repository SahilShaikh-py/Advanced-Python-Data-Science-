## 📌 1. Theory & Intuition
- **K-Means Clustering:** Iteratively updates $K$ cluster centroids to minimize within-cluster sum of squares (WCSS).
- **Principal Component Analysis (PCA):** Projects high-dimensional feature spaces onto orthogonal principal component axes maximizing variance.

```python
import numpy as np
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler

# 💻 Live Demo: K-Means Customer Segmentation & PCA 2D Projection
np.random.seed(42)
X_raw = np.random.rand(100, 6) # 100 customers, 6 behavioral features

# Scale Features
X_scaled = StandardScaler().fit_transform(X_raw)

# 1. K-Means Clustering (K=3)
kmeans = KMeans(n_clusters=3, random_state=42)
cluster_labels = kmeans.fit_predict(X_scaled)

# 2. PCA Dimensionality Reduction (6D -> 2D)
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)

print("Explained Variance Ratio by 2 Components:", pca.explained_variance_ratio_)
print("Total Variance Retained:", sum(pca.explained_variance_ratio_))
print("First 5 Cluster Assignments:", cluster_labels[:5])
```
