## 1. Definition
- Unsupervised Learning identifies hidden structural clusters in data without relying on target labels.
- PCA projects high-dimensional feature spaces onto orthogonal principal component axes.

## 2. Key Features
- K-Means minimizes Within-Cluster Sum of Squares (WCSS) to position K centroid seeds.
- Silhouette Score evaluates cluster separation quality between -1 and +1.
- Principal components capture directions of maximum dataset feature variance.
- Feature scaling (StandardScaler) is mandatory prior to distance computation.

## 3. Real-World Uses
- Customer behavioral segmentation for personalized marketing campaigns.
- Reducing feature dimensionality prior to complex model training.

```python
import numpy as np
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler

# Live Instructor Code-Along: K-Means Segmentation and PCA Projection
np.random.seed(42)
X_raw = np.random.rand(120, 6)

X_scaled = StandardScaler().fit_transform(X_raw)

# 1. K-Means Clustering
kmeans = KMeans(n_clusters=3, random_state=42)
cluster_labels = kmeans.fit_predict(X_scaled)

# 2. PCA Projection (6D to 2D)
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)

print("Explained Variance Ratio:", pca.explained_variance_ratio_)
print("Total Variance Retained:", round(sum(pca.explained_variance_ratio_), 4))
print("Cluster Assignments:", cluster_labels[:10])
```
