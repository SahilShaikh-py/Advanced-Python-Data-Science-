## 📌 Step 1: Load Dataset & Inspect Summary

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset
url = "https://raw.githubusercontent.com/SahilShaikh-py/Data_Analytic_/main/datasets/e_commerce_sales.csv"
df = pd.read_csv(url)
df["Total_Sales"] = df["Quantity"] * df["Unit_Price"]

print("Capstone Dataset Loaded Successfully!")
print(df.head())
```

## 📌 Step 2: City Revenue & Rating Analysis

```python
city_perf = df.groupby("City").agg(
    Total_Revenue=("Total_Sales", "sum"),
    Avg_Rating=("Rating", "mean"),
    Orders_Count=("Order_ID", "count")
).reset_index().sort_values(by="Total_Revenue", ascending=False)

print(city_perf)
```
