## 📌 Setup SQLite Database

```python
import sqlite3
import pandas as pd

conn = sqlite3.connect(":memory:")
url = "https://raw.githubusercontent.com/SahilShaikh-py/Data_Analytic_/main/datasets/e_commerce_sales.csv"
df_sales = pd.read_csv(url)
df_sales["Total_Sales"] = df_sales["Quantity"] * df_sales["Unit_Price"]
df_sales.to_sql("orders", conn, index=False, if_exists="replace")
print("Database loaded!")
```

## 📌 Topic 2.3: SQL Window Functions (`DENSE_RANK()`)

📌 **Definition:** Window Functions calculate ranks and metrics across row partitions without collapsing individual rows.

```python
query_window = """
SELECT Order_ID, City, Product, Total_Sales,
       DENSE_RANK() OVER (PARTITION BY City ORDER BY Total_Sales DESC) AS City_Rank
FROM orders;
"""

df_ranked = pd.read_sql_query(query_window, conn)
print(df_ranked.head(10))
```
