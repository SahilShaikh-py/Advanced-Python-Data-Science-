## 📌 Setup SQLite Database & Load GitHub CSV

```python
import sqlite3
import pandas as pd

# Create in-memory SQLite database connection
conn = sqlite3.connect(":memory:")

# Load sales data from GitHub raw URL into SQL table 'orders'
url = "https://raw.githubusercontent.com/SahilShaikh-py/Data_Analytic_/main/datasets/e_commerce_sales.csv"
df_sales = pd.read_csv(url)
df_sales["Total_Sales"] = df_sales["Quantity"] * df_sales["Unit_Price"]
df_sales.to_sql("orders", conn, index=False, if_exists="replace")

print("Database loaded successfully!")
```

## 📌 Topic 2.1: SQL Data Extraction (`SELECT`, `WHERE`, `ORDER BY`)

📌 **Definition:** SQL queries extract and filter specific rows from database tables.

📖 **Key Theory (5 Short Bullet Points):**
- 🔹 **`SELECT`:** Specifies columns to retrieve.
- 🔹 **`WHERE`:** Filters rows based on conditions.
- 🔹 **`ORDER BY`:** Sorts output rows `ASC` or `DESC`.
- 🔹 **`LIMIT`:** Restricts number of returned rows.
- 🔹 **Industry Application:** Querying high-value orders in specific cities.

```python
# Write SQL Query
query = """
SELECT Order_ID, City, Category, Product, Total_Sales, Rating
FROM orders
WHERE City = 'Mumbai' AND Total_Sales > 2000
ORDER BY Total_Sales DESC;
"""

df_result = pd.read_sql_query(query, conn)
print(df_result)
```

## 📌 Topic 2.2: Data Aggregation & Grouping (`GROUP BY`, `HAVING`)

📌 **Definition:** Grouping aggregates rows sharing common categorical attributes to calculate summary metrics.

```python
# SQL GroupBy Query
query_groupby = """
SELECT City, SUM(Total_Sales) AS City_Revenue, AVG(Rating) AS Avg_Rating, COUNT(Order_ID) AS Total_Orders
FROM orders
GROUP BY City
HAVING City_Revenue > 5000
ORDER BY City_Revenue DESC;
"""

df_summary = pd.read_sql_query(query_groupby, conn)
print(df_summary)
```
