## 📌 Topic 3.5: NumPy Vectorization vs Python Loops

📌 **Definition:** Vectorization executes operations on entire numerical arrays simultaneously without writing explicit `for` loops.

```python
import numpy as np

# Stock portfolio prices
stock_prices = np.array([120.50, 450.00, 890.25, 1250.00, 310.75])

# Instant Vectorized Calculation (5% fee inflation across all stocks)
adjusted_prices = stock_prices * 1.05

print("Original Stock Prices:", stock_prices)
print("Adjusted Prices (+5% Fee):", np.round(adjusted_prices, 2))
```
