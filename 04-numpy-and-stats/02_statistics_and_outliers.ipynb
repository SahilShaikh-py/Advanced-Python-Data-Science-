## 📌 Topic 3.6: Interquartile Range (IQR) Outlier Detection

📌 **Definition:** The IQR Method measures statistical spread ($Q3 - Q1$) to identify extreme values that fall outside expected limits.

```python
import numpy as np

# Transaction values with 1 fraud outlier (95,000)
orders = np.array([450, 500, 480, 520, 510, 490, 530, 470, 95000])

# Calculate Q1, Q3, IQR
q1 = np.percentile(orders, 25)
q3 = np.percentile(orders, 75)
iqr = q3 - q1
upper_limit = q3 + (1.5 * iqr)

outliers = orders[orders > upper_limit]
print(f"Q1: ₹{q1}, Q3: ₹{q3}, IQR: ₹{iqr}")
print(f"Upper Limit Threshold: ₹{upper_limit}")
print(f"Flagged Outliers: {outliers}")
```
