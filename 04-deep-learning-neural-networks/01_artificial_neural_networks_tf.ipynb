## 📌 1. Theory & Intuition
- **Forward Propagation:** Computing layer outputs $z = W X + b \rightarrow a = f(z)$.
- **Backpropagation:** Applying the chain rule to compute gradients $\frac{\partial L}{\partial W}$ and update neural weights.

```python
import tensorflow as tf
from tensorflow.keras import layers, models
import numpy as np

# 💻 Live Demo: Training a 3-Layer Dense Artificial Neural Network
X_train = np.random.rand(500, 10) # 500 samples, 10 features
y_train = np.random.choice([0, 1], size=500)

model = models.Sequential([
    layers.Dense(32, activation='relu', input_shape=(10,)),
    layers.Dropout(0.2),
    layers.Dense(16, activation='relu'),
    layers.Dense(1, activation='sigmoid')
])

model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
history = model.fit(X_train, y_train, epochs=10, batch_size=32, verbose=1)

print("Training Complete. Final Loss:", history.history['loss'][-1])
```
