## 1. Definition
- Convolutional Neural Networks extract spatial features from grid-structured image data using sliding weight filters.
- Pooling layers downsample spatial feature maps while preserving key activations.

## 2. Key Features
- Conv2D layers slide KxK weight filters across input image channels.
- Parameter sharing drastically reduces total trainable weight parameters compared to Dense layers.
- MaxPooling reduces spatial height and width dimensions, conferring translation invariance.
- Transfer learning reuses pre-trained deep CNN backbones (ResNet, MobileNet) on custom tasks.

## 3. Real-World Uses
- Medical X-ray and MRI diagnostic image classification.
- Autonomous vehicle obstacle detection and vision perception.

```python
import tensorflow as tf
from tensorflow.keras import layers, models

# Live Instructor Code-Along: Conv2D Image Classification Architecture
cnn_model = models.Sequential([
    layers.Conv2D(32, (3, 3), activation='relu', input_shape=(128, 128, 3)),
    layers.MaxPooling2D((2, 2)),
    
    layers.Conv2D(64, (3, 3), activation='relu'),
    layers.MaxPooling2D((2, 2)),
    
    layers.Flatten(),
    layers.Dense(64, activation='relu'),
    layers.Dense(1, activation='sigmoid')
])

cnn_model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
cnn_model.summary()
```
