## 📌 Topic 4.1: Reading CSVs & Inspecting DataFrames

📌 **Definition:** Pandas DataFrames store tabular data in rows and columns with powerful analytical methods.

```python
import pandas as pd

# Load dataset from GitHub raw URL
url = "https://raw.githubusercontent.com/SahilShaikh-py/Data_Analytic_/main/datasets/e_commerce_sales.csv"
df = pd.read_csv(url)

print("=== First 5 Rows ===")
print(df.head())

print("\n=== Summary Info ===")
print(f"Shape: {df.shape[0]} Rows, {df.shape[1]} Columns")
```

## 📌 Topic 4.2: Conditional Data Filtering (`.loc[]`)

📌 **Definition:** Filtering selects specific subset rows based on boolean logical conditions.

```python
# Filter Electronics items in Mumbai
mumbai_electronics = df[(df["City"] == "Mumbai") & (df["Category"] == "Electronics")]
print(mumbai_electronics[["Order_ID", "Product", "Quantity", "Unit_Price"]])
```
