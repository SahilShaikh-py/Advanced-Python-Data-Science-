## 📌 Topic 4.3: GroupBy & Multi-Metric Aggregation

📌 **Definition:** GroupBy splits data into categorical groups, applies mathematical aggregations, and combines results.

```python
import pandas as pd

url = "https://raw.githubusercontent.com/SahilShaikh-py/Data_Analytic_/main/datasets/e_commerce_sales.csv"
df = pd.read_csv(url)
df["Total_Sales"] = df["Quantity"] * df["Unit_Price"]

# Multi-metric summary by City
city_summary = df.groupby("City").agg(
    Total_Revenue=("Total_Sales", "sum"),
    Average_Order=("Total_Sales", "mean"),
    Total_Orders=("Order_ID", "count")
).reset_index().sort_values(by="Total_Revenue", ascending=False)

print(city_summary)
```
