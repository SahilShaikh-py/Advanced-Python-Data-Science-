## 1. Definition
- Text Embeddings convert raw textual strings into dense floating-point vector representations capturing semantic meaning.
- Vector Databases index high-dimensional embeddings for sub-millisecond similarity search.

## 2. Key Features
- Tokenization breaks unstructured text into numerical token IDs.
- Dense Embeddings preserve semantic context (e.g. King - Man + Woman = Queen).
- Cosine Similarity measures the angular alignment between embedding vectors.
- Vector DBs use HNSW indexing to query top-k nearest neighbors in sub-milliseconds.

## 3. Real-World Uses
- Enterprise document search and knowledge retrieval.
- Context retrieval for Retrieval-Augmented Generation (RAG) AI Chatbots.

```python
import numpy as np
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

# Live Instructor Code-Along: TF-IDF Embedding Vector Search
documents = [
    "Data science and machine learning with Python",
    "Artificial intelligence and deep learning neural networks",
    "Delicious Italian pizza and pasta recipes"
]

vectorizer = TfidfVectorizer()
tfidf_matrix = vectorizer.fit_transform(documents)

similarity_scores = cosine_similarity(tfidf_matrix[0:1], tfidf_matrix)
print("Query Document: 'Data science and machine learning with Python'")
print("Similarity Scores across documents:", np.round(similarity_scores, 4))
```
