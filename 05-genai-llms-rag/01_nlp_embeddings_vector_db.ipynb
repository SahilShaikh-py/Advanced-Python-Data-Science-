## 📌 1. Theory & Intuition
- **Text Embeddings:** Converting unstructured text into dense floating-point vectors capturing contextual semantic meaning.
- **Vector Databases:** Indexing high-dimensional embeddings for sub-millisecond similarity search using Cosine Distance.

```python
import numpy as np
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

# 💻 Live Demo: Text Vectorization & Semantic Cosine Similarity
documents = [
    "Data science and machine learning with Python",
    "Artificial intelligence and deep learning neural networks",
    "Delicious Italian pizza and pasta recipes"
]

vectorizer = TfidfVectorizer()
tfidf_matrix = vectorizer.fit_transform(documents)

similarity_matrix = cosine_similarity(tfidf_matrix[0:1], tfidf_matrix)
print("Query Document: 'Data science and machine learning with Python'")
print("Similarity Scores:", similarity_matrix)
```
