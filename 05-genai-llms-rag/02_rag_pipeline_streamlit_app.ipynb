## 1. Definition
- Retrieval-Augmented Generation (RAG) dynamically retrieves private domain context to ground Large Language Model (LLM) responses.

## 2. Key Features
- Document Chunking breaks large text files into overlapping token chunks.
- Vector DB Indexing stores text embeddings for sub-millisecond context retrieval.
- Prompt Augmentation injects retrieved text chunks into the system prompt context window.
- Hallucination Reduction forces LLMs to generate factual answers grounded strictly in retrieved context.

## 3. Real-World Uses
- Enterprise HR policy and financial PDF document Q&A Chatbots.
- Domain-specific customer support automation.

```python
# Live Instructor Code-Along: Modular RAG Context Retriever Function

def rag_retriever_pipeline(user_query, document_chunks, top_k=2):
    # Simulated RAG Vector Search & Prompt Construction
    print(f"[RAG SEARCH] Query: '{user_query}'")
    print(f"[RETRIEVAL] Retrieving top {top_k} knowledge chunks...")
    
    retrieved_context = " \n".join(document_chunks[:top_k])
    augmented_prompt = f"System: Answer the question using ONLY the provided context.\nContext:\n{retrieved_context}\n\nQuestion: {user_query}"
    return augmented_prompt

chunks = [
    "Company return policy permits full refunds within 30 days of purchase.",
    "Customer support operates 24/7 via live email and chat.",
    "International shipping takes 5-7 business days."
]

prompt = rag_retriever_pipeline("What is the refund window?", chunks, top_k=1)
print("
=== Constructed Augmented Prompt ===
", prompt)
```
