## 📌 1. Theory & Intuition
- **RAG Pipeline:** Document Chunking $\rightarrow$ Vector Embedding $\rightarrow$ Vector DB Indexing $\rightarrow$ User Query $\rightarrow$ Context Retrieval $\rightarrow$ LLM Prompt Augmentation.
- **Hallucination Prevention:** Grounding LLM responses strictly within retrieved domain document chunks.

```python
# 💻 Live Demo: Modular RAG Context Retriever Function
def rag_retriever_pipeline(user_query, document_chunks, embeddings_matrix):
    # Simulated RAG Retrieval Logic
    print(f"[RAG SEARCH] Query: '{user_query}'")
    print(f"[RETRIEVAL] Found {len(document_chunks)} relevant knowledge chunks.")
    
    context = " \n".join(document_chunks[:2])
    prompt = f"System: Answer the query based ONLY on context.\nContext: {context}\nQuery: {user_query}"
    return prompt

chunks = [
    "Company refund policy allows returns within 30 days of purchase.",
    "Customer support operates 24/7 via live email and chat."
]
augmented_prompt = rag_retriever_pipeline("What is the return policy?", chunks, None)
print("
Constructed Augmented Prompt:
", augmented_prompt)
```
