## 📌 Topic 4.4: Category Revenue Bar Charts & Styling

📌 **Definition:** Visual graphics transform raw numbers into intuitive charts for business decision making.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

url = "https://raw.githubusercontent.com/SahilShaikh-py/Data_Analytic_/main/datasets/e_commerce_sales.csv"
df = pd.read_csv(url)
df["Total_Sales"] = df["Quantity"] * df["Unit_Price"]

# Bar chart
plt.figure(figsize=(8, 4))
sns.barplot(data=df, x="Category", y="Total_Sales", estimator=sum, errorbar=None, palette="viridis")
plt.title("Total Sales Revenue by Product Category", fontsize=12, fontweight="bold")
plt.ylabel("Revenue (₹)")
plt.tight_layout()
plt.show()
```
