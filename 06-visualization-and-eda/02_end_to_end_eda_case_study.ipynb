## 📌 Topic 4.5: Seaborn Correlation Matrix Heatmap

📌 **Definition:** Correlation heatmaps display color-coded Pearson correlation coefficients ($r$) between numerical columns.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

url = "https://raw.githubusercontent.com/SahilShaikh-py/Data_Analytic_/main/datasets/e_commerce_sales.csv"
df = pd.read_csv(url)
df["Total_Sales"] = df["Quantity"] * df["Unit_Price"]

# Correlation Heatmap
plt.figure(figsize=(6, 4))
sns.heatmap(df[["Quantity", "Unit_Price", "Rating", "Total_Sales"]].corr(), annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Correlation Matrix Heatmap")
plt.tight_layout()
plt.show()
```
